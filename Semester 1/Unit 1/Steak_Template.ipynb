## My Question
### Are people who drink take more risky bets then those who don't drink?

## My Study
### My study is an observational study because it measures and compares existing behaviors (whether people drink alcohol and which lottery bet they choose) without applying any experimental treatment, intervention, or random assignment.

### My Analysis

```python
import numpy as np
import pandas as pd
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('steak.csv')
```

```python
lottery_col = df.columns[1]  
   
drink_col = df.columns[3]
```

```python

df_clean = df.dropna(subset=[lottery_col, drink_col])


```

```python
drinkers = df_clean[df_clean[drink_col] == "Yes"]
non_drinkers = df_clean[df_clean[drink_col] == "No"]
```

```python
drinkers_rel = drinkers[lottery_col].value_counts(normalize=True).round(3)
non_drinkers_rel = non_drinkers[lottery_col].value_counts(
    normalize=True
).round(3)
```

```python
comparison_df = pd.DataFrame(
    {"Drinkers": drinkers_rel, "Non-Drinkers": non_drinkers_rel}
)

print("--- Relative Frequencies ---")
print(comparison_df)
```
