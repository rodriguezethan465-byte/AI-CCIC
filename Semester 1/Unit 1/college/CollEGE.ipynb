## My question:

### How much bigger is the z-score of Stanfords average SAT score compared to the z score of Fort Lewis average SAT score in our data?

## My study:

### My study is an census study using data on the college scorecard website. which is maintained by the US department of education and has data for all secondary insitutins in the US. so its a population based investigation.

## My analysis:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('cleaned_college.csv')
plot_df = df.dropna(subset=["sat_average"]).copy()
```

```python
print("Number of SAT rows:", df["sat_average"].dropna().count())
stanford_sat = plot_df[
    plot_df["school_name"].str.contains("Stanford", na=False)
]["sat_average"].values[0]
fort_lewis_sat = plot_df[
    plot_df["school_name"].str.contains("Fort Lewis", na=False)
]["sat_average"].values[0]
```

```python
df = pd.read_csv("cleaned_college.csv")
plot_data = df["sat_average"].dropna()

# Calculate stats
mean_val = plot_data.mean()
median_val = plot_data.median()

# Plot
fig, ax = plt.subplots(figsize=(10, 6))
sns.histplot(plot_data, color="steelblue", bins=30, ax=ax)

ax.axvline(mean_val, color="crimson", linewidth=2, label=f"Mean ({mean_val:.1f})")
ax.axvline(
    median_val,
    color="darkorange",
    linestyle="--",
    linewidth=2,
    label=f"Median ({median_val:.1f})",
)
plt.axvline(
    x=stanford_sat,
    color="purple",
    linestyle=":",
    linewidth=2,
    label=f"Stanford Avg SAT ({int(stanford_sat)})",
)
plt.axvline(
    x=fort_lewis_sat,
    color="teal",
    linestyle=":",
    linewidth=2,
    label=f"Fort Lewis Avg SAT ({int(fort_lewis_sat)})",
)
ax.set_title("Distribution of Average SAT Scores")
ax.set_xlabel("Average SAT Score")
ax.set_ylabel("Count")
ax.legend()

plt.show()
```

```python

sat_mean = plot_df["sat_average"].mean()
sat_std = plot_df["sat_average"].std()
print(sat_std)
print(sat_mean)
sat_list = plot_df["sat_average"].dropna().astype(int).tolist()

# Print formatted as a Desmos list
print("L = " + str(sat_list))
plot_df["sat_zscore"] = (plot_df["sat_average"] - sat_mean) / sat_std


schools = plot_df[
    plot_df["school_name"].str.contains("Stanford|Fort Lewis", na=False)
]
schools[["school_name", "sat_average", "sat_zscore"]]
```
