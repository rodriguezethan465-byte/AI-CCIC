# My question

### Do songs before 2020 get more streams then songs after 2020?

## *My study*

### My study is a obsevational study because this question requires data across two groups old songs and new songs both groups within a sample proving why my study is obsevational. 

## My analysis:

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv('spotify-2023.csv', encoding="latin-1")
```

```python
df['streams'] = pd.to_numeric(df['streams'], errors='coerce')

# 2. Create group labels based on release year
df['era'] = df['released_year'].apply(lambda yr: 'Before 2020' if yr < 2020 else '2020 or Later')

# 3. Filter out missing stream values
plot_df = df.dropna(subset=['streams'])
```

```python
plt.figure(figsize=(10, 6))
```

```python
sns.histplot(
    data=plot_df,
    x='streams',
    hue='era',
    element='step',
    stat='density',
    common_norm=False,
    bins=30,
    palette={'Before 2020': 'navy', '2020 or Later': 'orange'}
)
median_old = plot_df[plot_df['era'] == 'Before 2020']['streams'].median()
median_new = plot_df[plot_df['era'] == '2020 or Later']['streams'].median()

plt.axvline(median_old, color='navy', linestyle='--', linewidth=2, label=f'Median Before 2020 ({median_old:,.0f})')
plt.axvline(median_new, color='orange', linestyle='--', linewidth=2, label=f'Median 2020+ ({median_new:,.0f})')
```

```python


# Load dataset and clean streams
df = pd.read_csv('spotify-2023.csv', encoding='latin-1')
df['streams'] = pd.to_numeric(df['streams'], errors='coerce')

# Separate groups
pre_2020 = df[df['released_year'] < 2020]['streams'].dropna().tolist()
post_2020 = df[df['released_year'] >= 2020]['streams'].dropna().tolist()

# Print Desmos-ready list format
print(f"A = {pre_2020}")
print(f"B = {post_2020}")
```
